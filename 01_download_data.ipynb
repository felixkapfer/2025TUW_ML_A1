```python
import kaggle
import zipfile
```

# Download the Datasets

## Start downloading the loan dataset

First we download the loan dataset using the kaggle api. This will download a zip file to the following path

'data/downloads/'

After downloading the zip, we need to extract the files that the zip contains into the following folder:

'data/ds_loan'

```python
# 1. Download loan dataset from Kaggle using the Kaggle API
print("Downloading loan dataset from Kaggle...")
kaggle.api.competition_download_files('184-702-tu-ml-2025-w-loan', 
                                       path='./data/downloads', 
                                       quiet=False)


# 2. Extract the downloaded zip file
print("Extracting loan dataset...")
with zipfile.ZipFile('./data/downloads/184-702-tu-ml-2025-w-loan.zip', 'r') as zip_ref:
    zip_ref.extractall('./data/ds_loan')

print("Dataset downloaded and extracted successfully.")
```

## Start downloading the Breast Cancer Dataset

After downloading the loan dataset, we also need to download the second dataset containing Breast Cancer data.

To do so, we will also use the Kaggle API.

```python
# Download Breast Cancer Dataset from Kaggle using the Kaggle API
print("Downloading Breast Cancer dataset from Kaggle...")
kaggle.api.competition_download_files('184-702-tu-ml-2025w-breast-cancer',
                                        path='./data/downloads', 
                                        quiet=False)


# Extract the downloaded zip file
print("Extracting Breast Cancer dataset...")
with zipfile.ZipFile('./data/downloads/184-702-tu-ml-2025w-breast-cancer.zip', 'r') as zip_ref:
    zip_ref.extractall('./data/ds_breast_cancer')

print("Dataset downloaded and extracted successfully.")
```
